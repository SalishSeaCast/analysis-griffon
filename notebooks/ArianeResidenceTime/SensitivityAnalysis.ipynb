## Sensitivity Analysis Plotting of Tracer Filters

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
res_times_filters = pd.read_excel('/ocean/ghoeven/MOAD/Data/SensitivitySummary.xlsx', index_col=0)
res_times_filters
```

```python
fig, ax = plt.subplots(1,1, figsize=(4, 4))

labels=[]
i=0
marker_shapes = ['d', 's', '*', '1', 'o']
for col in res_times_filters:
    ax.scatter(list(res_times_filters.index), res_times_filters[col], s=50, marker=marker_shapes[i])
    i+=1
    labels += [col]

ax.set_ylabel('Residence Time (d)')
ax.set_xticklabels(list(res_times_filters.index), rotation=30)
ax.legend(labels, title='Filter Used', bbox_to_anchor=(1,1))
```
